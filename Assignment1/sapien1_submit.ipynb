```python
import zipfile
import os
from tqdm import tqdm

zip_path = '/content/drive/MyDrive/pascal_voc.zip'
extract_path = '/content/drive/MyDrive/pascal_voc'

# Create extraction directory if it doesn't exist
os.makedirs(extract_path, exist_ok=True)

# Get total number of files in zip
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    total_files = len(zip_ref.namelist())

# Extract with progress bar
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    for file in tqdm(zip_ref.namelist(), desc="Extracting", unit="file", total=total_files):
        zip_ref.extract(file, extract_path)

print("✅ Extraction complete!")
```

```python
import random, cv2, torch, os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from torchvision.models.detection import fasterrcnn_resnet50_fpn
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor
from torchvision.transforms import functional as F

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

DATASET_ROOT = "/content/drive/MyDrive/dataset1/pascal_voc/VOC2012_test/VOC2012_test"
MODEL_PATH = "/content/drive/MyDrive/fasterrcnn_pretrained.pth"

CLASSES = ["background","person","car","bicycle","dog","chair"]

# Build model architecture again
num_classes = len(CLASSES)
model = fasterrcnn_resnet50_fpn(weights=None)
in_features = model.roi_heads.box_predictor.cls_score.in_features
model.roi_heads.box_predictor = FastRCNNPredictor(in_features, num_classes)

# Load trained weights
model.load_state_dict(torch.load(MODEL_PATH, map_location=device))
model.to(device)
model.eval()

# Visualization function
def visualize(img, preds, score_thr=0.6):
    img = np.array(img)
    for box, label, score in zip(preds["boxes"], preds["labels"], preds["scores"]):
        if score < score_thr: continue
        x1,y1,x2,y2 = map(int, box.tolist())
        cv2.rectangle(img,(x1,y1),(x2,y2),(0,255,0),2)
        name = CLASSES[label]
        cv2.putText(img, f"{name} {score:.2f}", (x1,y1-5),
                    cv2.FONT_HERSHEY_SIMPLEX,0.6,(255,0,0),2)
    return img

# Pick 5 random images
image_files = os.listdir(os.path.join(DATASET_ROOT,"JPEGImages"))
sample_images = random.sample(image_files, 5)

plt.figure(figsize=(15,10))

for i, img_name in enumerate(sample_images, 1):
    path = os.path.join(DATASET_ROOT,"JPEGImages", img_name)
    pil = Image.open(path).convert("RGB")
    tensor = F.to_tensor(pil).unsqueeze(0).to(device)

    with torch.no_grad():
        preds = model(tensor)[0]

    result = visualize(pil, preds)
    plt.subplot(2,3,i)
    plt.imshow(result)
    plt.title(img_name)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
