```python

#  PCB DEFECT INSPECTION SYSTEM PIPELINE


!pip install -q ultralytics opencv-python tqdm scikit-learn

import torch, cv2, os, random, shutil, yaml
import numpy as np
from tqdm import tqdm
from sklearn.model_selection import train_test_split
from ultralytics import YOLO
import matplotlib.pyplot as plt


#  CUDA SETUP

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print("Using device:", device)


#  PATH CONFIG

BASE_PATH = "/content/drive/MyDrive/dataset2/pcb/DeepPCB-master/PCBData"

STAGING = "/content/pcb_staging"      # raw converted data
DATASET = "/content/pcb_dataset"      # final YOLO dataset
DRIVE_SAVE = "/content/drive/MyDrive/pcb_yolo_model"

for p in [STAGING, DATASET, DRIVE_SAVE]:
    os.makedirs(p, exist_ok=True)


#  DEFECT MAP

defect_map = {1:"Open",2:"Short",3:"MouseBite",4:"Spur",5:"PinHole",6:"Spurious"}


#  CONVERT DeepPCB → YOLO

print("Converting dataset...")

img_stage = f"{STAGING}/images"
lbl_stage = f"{STAGING}/labels"
os.makedirs(img_stage, exist_ok=True)
os.makedirs(lbl_stage, exist_ok=True)

for group in tqdm(os.listdir(BASE_PATH)):
    gpath = os.path.join(BASE_PATH, group)
    if not os.path.isdir(gpath): continue

    img_folder = [f for f in os.listdir(gpath) if not f.endswith("_not")][0]
    lbl_folder = img_folder + "_not"

    for imgfile in os.listdir(f"{gpath}/{img_folder}"):
        if not imgfile.endswith("_test.jpg"): continue

        name = imgfile.replace("_test.jpg","")
        img_src = f"{gpath}/{img_folder}/{imgfile}"
        lbl_src = f"{gpath}/{lbl_folder}/{name}.txt"

        img = cv2.imread(img_src)
        h,w,_ = img.shape

        yolo_lbl = []
        if os.path.exists(lbl_src):
            with open(lbl_src) as f:
                for line in f:
                    x1,y1,x2,y2,cls = map(int,line.split())
                    xc = (x1+x2)/2/w
                    yc = (y1+y2)/2/h
                    bw = (x2-x1)/w
                    bh = (y2-y1)/h
                    yolo_lbl.append(f"{cls-1} {xc} {yc} {bw} {bh}")

        shutil.copy(img_src, f"{img_stage}/{name}.jpg")
        with open(f"{lbl_stage}/{name}.txt","w") as f:
            f.write("\n".join(yolo_lbl))


#  TRAIN / VAL SPLIT

images = os.listdir(img_stage)
train, val = train_test_split(images, test_size=0.2, random_state=42)

for s in ["train","val"]:
    os.makedirs(f"{DATASET}/images/{s}", exist_ok=True)
    os.makedirs(f"{DATASET}/labels/{s}", exist_ok=True)

for x in train:
    shutil.move(f"{img_stage}/{x}", f"{DATASET}/images/train/{x}")
    shutil.move(f"{lbl_stage}/{x.replace('.jpg','.txt')}", f"{DATASET}/labels/train/{x.replace('.jpg','.txt')}")

for x in val:
    shutil.move(f"{img_stage}/{x}", f"{DATASET}/images/val/{x}")
    shutil.move(f"{lbl_stage}/{x.replace('.jpg','.txt')}", f"{DATASET}/labels/val/{x.replace('.jpg','.txt')}")


#  DATA YAML

data_yaml = {
    'path': DATASET,
    'train': 'images/train',
    'val': 'images/val',
    'names': {i:defect_map[i+1] for i in range(6)}
}

with open(f"{DATASET}/pcb.yaml",'w') as f:
    yaml.dump(data_yaml,f)


#  TRAIN MODEL

model = YOLO("yolov8s.pt").to(device)
model.train(data=f"{DATASET}/pcb.yaml", epochs=25, imgsz=640, device=device)


#  SAVE MODEL TO GOOGLE DRIVE

best_model_path = "runs/detect/train/weights/best.pt"
shutil.copy(best_model_path, f"{DRIVE_SAVE}/pcb_defect_yolo.pt")
print("Model saved:", f"{DRIVE_SAVE}/pcb_defect_yolo.pt")


#  FINAL INFERENCE DEMO

model = YOLO(f"{DRIVE_SAVE}/pcb_defect_yolo.pt").to(device)
samples = random.sample(os.listdir(f"{DATASET}/images/val"), 5)

def severity(area):
    if area < 0.001: return "Low"
    if area < 0.01: return "Medium"
    return "High"

for imgname in samples:
    img = cv2.imread(f"{DATASET}/images/val/{imgname}")
    h,w,_ = img.shape
    results = model(img, device=device)[0]

    print(f"\n📸 {imgname}")
    for b in results.boxes:
        cls = int(b.cls[0])
        conf = float(b.conf[0])
        x1,y1,x2,y2 = map(int,b.xyxy[0])

        xc = int((x1+x2)/2)
        yc = int((y1+y2)/2)
        area = (x2-x1)*(y2-y1)/(w*h)
        sev = severity(area)

        print(f"Defect: {defect_map[cls+1]} | Conf:{conf:.2f} | Center:({xc},{yc}) | Severity:{sev}")

        cv2.rectangle(img,(x1,y1),(x2,y2),(0,255,0),2)
        cv2.putText(img,f"{defect_map[cls+1]} {conf:.2f}",(x1,y1-6),
                    cv2.FONT_HERSHEY_SIMPLEX,0.6,(0,255,0),2)

    plt.figure(figsize=(6,6))
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()
```
